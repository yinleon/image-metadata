# Video Metadata

We'll recreate how Gizmodo and ProPublica verified videos of protesters inside the U.S. Capitol Building for the shories "[Parler Users Breached Deep Inside U.S. Capitol Building, GPS Data Shows](https://gizmodo.com/parler-users-breached-deep-inside-u-s-capitol-building-1846042905)" and "[What Parler Saw During the Attack on the Capitol](https://projects.propublica.org/parler-capitol-videos/)" published Jan. 2021.

Download exiftool and the Python wrapper. Read [documentation here](https://pypi.org/project/PyExifTool/#getting-pyexiftool).

We'll look at a video from the January 6 insurrection as recorded on Parler. Read the documentation of this dataset by [DDoSecrets](https://ddosecrets.com/article/parler).

```python
import os
import urllib.request
from IPython.display import Video
import exiftool
```

```python
video_id = 'AYijYnzI0Gbr'
url = f'https://data.ddosecrets.com/Parler/Videos/{video_id}'
fn_video = f'data/{video_id}.mp4'

if not os.path.exists(fn_video):
    urllib.request.urlretrieve(url, fn_video)
```

```python
Video(fn_video)
```

## looking at metadata

We will use `exiftool` to see when the video was created, whether it was shot on mobile, and where the video was taken.

Note: Theese metadata fields and their names differ depending on manufacturer of the camera/phone. So you some logic is needed to choose the correct fields based on the whatever shot the video. Additionally, users can turn off location tracking on photos and videos.

The code below works for iPhones:

```python
import exiftool

with exiftool.ExifToolHelper() as et:
    metadata = et.get_metadata(fn_video)
```

```python
metadata
```

```python
vid_data = metadata[0]
```

```python
# when was the video recorded?
vid_data['QuickTime:CreationDate']
```

```python
# was the video shot on mobile?
vid_data['QuickTime:Make'], vid_data['QuickTime:Model']
```

```python
# where was the video taken?
lat, lon = vid_data['Composite:GPSLatitude'], vid_data['Composite:GPSLongitude']
```
